## Activity 1: Create and Inspect DataFrame
We will create a DataFrame with 10 students including their Name, Age, City, Attendance, and Marks.

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ivan', 'Jack'],
    'Age': [20, 21, 19, 22, 20, 23, 21, 20, 22, 19],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Chicago', 'New York', 'Miami', 'Los Angeles', 'Houston'],
    'Attendance': [95, 88, 70, 92, 85, 60, 98, 77, 89, 94],
    'Marks': [85, 76, 55, 90, 82, 45, 95, 68, 88, 91]
}

df = pd.DataFrame(data)

print("--- Index ---")
print(df.index)
print("\n--- Columns ---")
print(df.columns)
display(df)
```

## Activity 2: Exploratory Data Analysis (EDA)
We will use various pandas methods to inspect the dataset and write observations.

```python
print("--- head(2) ---")
display(df.head(2))

print("\n--- tail(2) ---")
display(df.tail(2))

print("\n--- info() ---")
df.info()

print("\n--- describe() ---")
display(df.describe())

print("\n--- sample(3) ---")
display(df.sample(3))

print("\n--- value_counts() for City ---")
print(df['City'].value_counts())

print("\n--- unique() Cities ---")
print(df['City'].unique())

print(f"\nShape: {df.shape}")
print(f"Size: {df.size}")
```

### Activity 2 Observations:
1. The dataset contains 10 entries and 5 columns.
2. The average mark across all students is approximately 78.
3. Chicago, New York, Los Angeles, and Houston appear multiple times in the 'City' column.
4. There are no missing values (all 10 are non-null according to info()).
5. Marks range from a minimum of 45 to a maximum of 95.

## Activity 3: Selection and Indexing
Using `loc` and `iloc` to retrieve specific parts of the data.

```python
# Select individual column
names = df['Name']

# Select multiple columns
subset = df[['Name', 'Marks']]

# Use iloc to get the first 3 rows and first 2 columns
iloc_selection = df.iloc[0:3, 0:2]

# Use loc to get rows where index is 0 to 2 for 'Name' and 'City'
loc_selection = df.loc[0:2, ['Name', 'City']]

print("--- iloc selection (0:3, 0:2) ---")
display(iloc_selection)

print("\n--- loc selection (0:2, ['Name', 'City']) ---")
display(loc_selection)
```

## Activity 4: Filtering Data
We will use logical operators to filter the dataset: `&` (AND), `|` (OR), `~` (NOT), and `^` (XOR).

```python
# 1. AND (&): Students from New York AND Marks > 90
filter_and = df[(df['City'] == 'New York') & (df['Marks'] > 90)]
print("--- AND Filter (NY & Marks > 90) ---")
display(filter_and)

# 2. OR (|): Students from Miami OR Chicago
filter_or = df[(df['City'] == 'Miami') | (df['City'] == 'Chicago')]
print("\n--- OR Filter (Miami | Chicago) ---")
display(filter_or)

# 3. NOT (~): Students NOT from Houston
filter_not = df[~(df['City'] == 'Houston')]
print("\n--- NOT Filter (~Houston) ---")
display(filter_not)

# 4. XOR (^): Students (Age < 20) XOR (Marks > 85)
# This selects students who meet exactly ONE of the two conditions
filter_xor = df[(df['Age'] < 20) ^ (df['Marks'] > 85)]
print("\n--- XOR Filter (Age < 20 ^ Marks > 85) ---")
display(filter_xor)
```

## Activity 5: Modifying the DataFrame
We will add a 'Result' column, adjust the 'Marks', drop 'Age', and perform sorting.

```python
# Add a Pass/Fail column based on Marks
df['Result'] = np.where(df['Marks'] >= 50, 'Pass', 'Fail')

# Modify a numerical column: Add a 2-point bonus to Attendance (capped at 100)
df['Attendance'] = (df['Attendance'] + 2).clip(upper=100)

# Drop an unnecessary column (e.g., Age)
df_modified = df.drop(columns=['Age'])

# Sort by Marks descending
print("--- Sorted by Marks (Descending) ---")
display(df_modified.sort_values(by='Marks', ascending=False))

# Sort by City ascending
print("\n--- Sorted by City (Ascending) ---")
display(df_modified.sort_values(by='City', ascending=True))
```

## Activity 6: Handling Missing Values
We will intentionally introduce NaNs, detect them, and handle them using `dropna()` and `fillna()`.

```python
# Intentionally creating missing values
df_missing = df.copy()
df_missing.loc[2, 'Marks'] = np.nan
df_missing.loc[5, 'Attendance'] = np.nan
df_missing.loc[8, 'Marks'] = np.nan

print("--- Detecting Missing Values ---")
print(df_missing.isnull().sum())

# Remove records with missing values
df_dropped = df_missing.dropna()
print(f"\nRows after dropna(): {len(df_dropped)}")

# Replace missing Marks with the mean
mean_marks = df_missing['Marks'].mean()
df_missing['Marks'] = df_missing['Marks'].fillna(mean_marks)

# Replace missing Attendance with the median
median_att = df_missing['Attendance'].median()
df_missing['Attendance'] = df_missing['Attendance'].fillna(median_att)

print("\n--- DataFrame after fillna() ---")
display(df_missing)
```

## Activity 7: Descriptive Statistics
Applying statistical functions to the 'Marks' column to understand the distribution.

```python
stats = {
    'Sum': df['Marks'].sum(),
    'Mean': df['Marks'].mean(),
    'Median': df['Marks'].median(),
    'Min': df['Marks'].min(),
    'Max': df['Marks'].max(),
    'Count': df['Marks'].count(),
    'Unique Count': df['Marks'].nunique(),
    'Std Dev': df['Marks'].std()
}

for stat, value in stats.items():
    print(f"{stat}: {value:.2f}")

print("\nInterpretation: The average score is 78.5, while the median is 83.5, suggesting a slight left skew due to lower scores like 45. The standard deviation of 16.5 indicates a moderate spread in student performance.")
```

## Activity 8: Grouping and Aggregation
We will use `groupby()` and `agg()` to calculate statistics for different cities and then perform a multi-column grouping.

```python
# Single column grouping with multiple statistics
city_stats = df.groupby('City').agg({
    'Marks': ['mean', 'max'],
    'Attendance': 'mean'
})
print("--- City Statistics ---")
display(city_stats)

# Adding a dummy 'Department' column for multi-column grouping
df['Department'] = ['CS', 'CS', 'Math', 'Math', 'CS', 'Math', 'CS', 'CS', 'Math', 'Math']
multi_group = df.groupby(['City', 'Department'])['Marks'].mean()
print("\n--- Multi-column Grouping (City & Department) ---")
display(multi_group)
```

## Activity 9: Pivot Table
Creating a pivot table to summarize the average Marks based on City and Result status.

```python
pivot = df.pivot_table(values='Marks', index='City', columns='Result', aggfunc='mean')
print("--- Pivot Table: Average Marks by City and Result ---")
display(pivot)
print("\nExplanation: The table shows the average marks for students in each city, categorized by whether they Passed or Failed. NaN indicates no students fell into that category for that city.")
```

## Activity 10: Joining DataFrames
Creating a second DataFrame and demonstrating Inner, Outer, Left, and Right joins.

```python
# Create a related DataFrame with Contact Info
contact_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Zoe'],
    'Email': ['alice@edu.com', 'bob@edu.com', 'char@edu.com', 'dave@edu.com', 'zoe@edu.com']
}
df_contact = pd.DataFrame(contact_data)

print("--- Inner Join ---")
display(pd.merge(df, df_contact, on='Name', how='inner'))

print("\n--- Outer Join ---")
display(pd.merge(df, df_contact, on='Name', how='outer'))

print("\n--- Left Join ---")
display(pd.merge(df, df_contact, on='Name', how='left'))

print("\n--- Right Join ---")
display(pd.merge(df, df_contact, on='Name', how='right'))

# Demonstrate join() using index
df_indexed = df.set_index('Name')
df_contact_indexed = df_contact.set_index('Name')
print("\n--- Join using .join() (Left) ---")
display(df_indexed.join(df_contact_indexed, how='left'))
```

## Activity 11: Matplotlib and Seaborn Visualizations
We will create a line plot, scatter plot, histogram, bar plot, heatmap, and box plot.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set figure size
plt.figure(figsize=(15, 10))

# 1. Line Plot: Student Marks Trend
plt.subplot(2, 3, 1)
plt.plot(df['Name'], df['Marks'], marker='o', color='b')
plt.title('Marks per Student')
plt.xticks(rotation=45)

# 2. Scatter Plot: Age vs Attendance
plt.subplot(2, 3, 2)
plt.scatter(df['Age'], df['Attendance'], color='r')
plt.title('Age vs Attendance')
plt.xlabel('Age')
plt.ylabel('Attendance')

# 3. Histogram: Marks Distribution
plt.subplot(2, 3, 3)
plt.hist(df['Marks'], bins=5, color='skyblue', edgecolor='black')
plt.title('Marks Distribution')

# 4. Bar Plot: Average Marks by City
plt.subplot(2, 3, 4)
sns.barplot(x='City', y='Marks', data=df, errorbar=None)
plt.title('Avg Marks by City')
plt.xticks(rotation=45)

# 5. Heatmap: Correlation Matrix
plt.subplot(2, 3, 5)
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')

# 6. Box Plot: Marks by Result
plt.subplot(2, 3, 6)
sns.boxplot(x='Result', y='Marks', data=df)
plt.title('Marks Box Plot')

plt.tight_layout()
plt.show()
```

## Activity 12: Integrated EDA Task
Loading a mock CSV dataset, cleaning, filtering, and summarizing with plots.

```python
from io import StringIO

# 1. Load data
csv_data = """ID,Category,Price,Quantity,Rating\n1,A,100,5,4.2\n2,B,150,,3.8\n3,A,,10,4.5\n4,C,200,2,3.9\n5,B,120,8,4.1\n6,A,90,12,3.5"""
sales_df = pd.read_csv(StringIO(csv_data))

# 2. Inspect and Clean
print("--- Missing Values Before Cleaning ---")
print(sales_df.isnull().sum())
sales_df['Price'] = sales_df['Price'].fillna(sales_df['Price'].median())
sales_df['Quantity'] = sales_df['Quantity'].fillna(0)

# 3. Filter and Derived Column
sales_df = sales_df[sales_df['Price'] > 95]
sales_df['Revenue'] = sales_df['Price'] * sales_df['Quantity']

# 4. Summarize
print("\n--- Sales Summary by Category ---")
print(sales_df.groupby('Category')['Revenue'].sum())

# 5. Visualizations
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))
sns.barplot(x='Category', y='Revenue', data=sales_df, ax=ax1)
ax1.set_title('Revenue by Category')
sns.scatterplot(x='Price', y='Rating', data=sales_df, ax=ax2)
ax2.set_title('Price vs Rating')
sales_df['Category'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=ax3)
ax3.set_title('Category Share')
plt.show()
```

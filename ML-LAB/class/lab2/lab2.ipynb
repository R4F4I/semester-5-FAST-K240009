## function approximation

- **input space** $(X)$: set of all possible inputs
- **output space** $(Y)$: set of all possible outputs or labels
- **target function** (): the unkowns true mapping form $X$ to $Y$
- **training dataset** (DTrain): examples $(x_i,y_i)$ use by the learner
- **hypothesis** $(h)$: a candidate function used to approx. the target function
- **hypothesis space** $(H)$: set of candidate h functions availble to the learner
- **prediction**: the output $h(x)$  produced by the learned hypothesis
